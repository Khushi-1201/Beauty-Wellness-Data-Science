```python
import pandas as pd
import numpy as np
from pathlib import Path

# Step 1: Load dataset
input_file = Path("/content/salon_appointments_dummy.csv")
output_file = Path("/content/cleaned_salon_appointments.csv")

df = pd.read_csv(input_file)

print("Original Dataset Shape:", df.shape)

# Step 2: Display first five rows
print("\nFirst 5 Rows:")
print(df.head())

# Step 3: Check data types
print("\nData Types:")
print(df.dtypes)

# Step 4: Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Step 5: Check duplicate records
print("\nExact Duplicate Rows:", df.duplicated().sum())

print(
    "Duplicate Appointment IDs:",
    df["appointment_id"].duplicated().sum()
)

# Step 6: Create a working copy
cleaned_df = df.copy()

# Step 7: Standardize text columns
text_columns = [
    "service_name",
    "customer_id",
    "staff_member_id"
]

for column in text_columns:
    cleaned_df[column] = (
        cleaned_df[column]
        .astype("string")
        .str.strip()
    )

# Step 8: Convert appointment dates
cleaned_df["appointment_date"] = pd.to_datetime(
    cleaned_df["appointment_date"],
    format="%d-%m-%Y",
    errors="coerce"
)

# Step 9: Standardize appointment times
cleaned_df["appointment_time"] = pd.to_datetime(
    cleaned_df["appointment_time"],
    format="%H:%M",
    errors="coerce"
).dt.strftime("%H:%M")

# Step 10: Convert prices to numeric
cleaned_df["price"] = pd.to_numeric(
    cleaned_df["price"],
    errors="coerce"
)

# Step 11: Check missing values after conversion
print("\nMissing Values After Conversion:")
print(cleaned_df.isnull().sum())

# Step 12: Identify invalid prices
invalid_prices = (
    cleaned_df["price"].isna()
    | (cleaned_df["price"] <= 0)
)

print("\nInvalid Prices:", invalid_prices.sum())

# Step 13: Detect potential outliers
Q1 = cleaned_df["price"].quantile(0.25)
Q3 = cleaned_df["price"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

potential_outliers = cleaned_df[
    (cleaned_df["price"] < lower_bound)
    | (cleaned_df["price"] > upper_bound)
]

print("\nPotential Price Outliers:")
print(potential_outliers[["service_name", "price"]])

# Step 14: Save cleaned dataset
cleaned_df.to_csv(output_file, index=False)

# Step 15: Validate final dataset
print("\nFinal Dataset Shape:", cleaned_df.shape)

print(
    "Final Duplicate Rows:",
    cleaned_df.duplicated().sum()
)

print("\nCleaning Process Completed Successfully!")
print("Saved File:", output_file)
```
